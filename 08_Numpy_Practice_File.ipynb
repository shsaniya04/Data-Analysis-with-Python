```python
# Arithmetic Operations & Broadcasting
a = np.array([1, 2, 3, 4])
```

```python
# Comparison Operations
arr = np.array([12, 45, 7, 23, 90, 18])
```

```python
print(arr[arr > 20])
```

```python
print((arr > 20).sum())
```

```python
arr[arr < 20] = 0
arr
```

```python
# Dot Product & Matrix Multiplication
a = np.array([1,2,3])
b = np.array([4,5,6])
```

```python
np.dot(a, b)
```

```python
a @ b
```

```python
np.matmul(a, b)
```

```python
# Reshaping & Concatenation
a = np.array([1,2,3,4,5,6])
```

```python
a.reshape(3,2)
```

```python
a.reshape(2, 3)
```

```python
a = a.reshape(2, 3)
```

```python
b = np.array([[7,8,9],[10,11,12]])
```

```python
np.concatenate((a, b), axis = 0)
```

```python
a = np.array([1,2,3,4,5,6])
a = a.reshape(3, 2)
a
```

```python
b = b.reshape(3, 2)
b
```

```python
np.concatenate((a, b), axis = 0)
```
