# Branching using Conditional Statements in Python

### The `if` statement
* In Python, branching is implemented using the if statement, which is written as follows:

```
if condition:
    statement1
    statement2 
```

````python
# Check if a number is positive.
num = int(input("Enter an Integer:"))

if num > 0:
    print("Positive Integer: {}".format(num))
````

````python
# Check if a number is even.
num = int(input("Enter an Integer:"))

if num % 2 == 0:
    print("The number is even: {}".format(num))
````

````python
# Given age, check if a person is eligible to vote (>=18).
num = int(input("Enter your age:"))

if num >= 18:
    print("You are ELIGIBLE to vote!!")
````

````python
# Check if a character is a vowel (a, e, i, o, u).
vowel = ['a', 'e', 'i', 'o', 'u']
character = str(input("Enter a character: "))

if character in vowel:
    print("The character is a vowel.")
````

````python
# Check if a number is divisible by 3.
a_num = 27
if a_num % 3 == 0:
    print("We are inside an if block!")
    print("The given number {} is divisible by 3".format(a_num))
````

### The `if` with `else` statement

We may want to print a different message if the number is not even in the above example. This can be done by adding the `else` statement. It is written as follows:

```
if condition:
    statement1
    statement2
else:
    statement4
    statement5

```

If `condition` evaluates to `True`, the statements in the `if` block are executed. If it evaluates to `False`, the statements in the `else` block are executed.

````python
# Check if a number is even or odd.
num = int(input("Enter an Integer: "))

if num % 2 == 0:
    print("The given integer {} is even.".format(num))
else:
    print("The given integer {} is odd.".format(num))
````

### The `elif` statement

Python also provides an `elif` statement (short for "else if") to chain a series of conditional blocks. The conditions are evaluated one by one. For the first condition that evaluates to `True`, the block of statements below it is executed. The remaining conditions and statements are not evaluated. So, in an `if`, `elif`, `elif`... chain, at most one block of statements is executed, the one corresponding to the first condition that evaluates to `True`. 

````python
# Grade System
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade: A")
elif 89 >= marks >= 75:
    print("Grade: B")
elif 74 >= marks >= 50:
    print("Grade: C")
else:
    print("Grade: Fail")
````

````python
# Find the largest of three numbers
a = int(input("Enter First Integer : "))
b = int(input("Enter Second Integer : "))
c = int(input("Enter Third Integer : "))

if a > b and a > c:
    print("{} is Largest".format(a))
elif b > a and b > c:
    print("{} is Largest".format(b))
else: 
    print("{} is Largest".format(c))
````

````python
# Check if a number is divisible by both 3 and 5.
num = int(input("Enter an Integer: "))
div_3 = (num % 3 == 0)
div_5 = (num % 5 == 0)
if div_3 and div_5:
    print("The number {} is divisible by both 3 and 5".format(num))
elif div_3 == True:
    print("The number {} is divisible by 3.".format(num))
else:
    print("The number {} is divisible by 5.".format(num))
````

````python
# Simple calculator using operator input (+, -, *, /).
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    print("Result: ", num1 + num2)
elif operator == "-":
    print("Result: ", num1 - num2)
elif operator == "*":
    print("Result: ", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result: ", num1 / num2)
    else: 
        print("Error: Division by zero")
else:
    print("Invalid operator.")
````

### Logical Thinking (Real Brain Work)

````python
# Check if a number is a multiple of 7 but not 5.
num1 = int(input("Enter an Integer: "))
multiple_7 = (num1 % 7 == 0)
multiple_5 = (num1 % 5 ==0)

if multiple_7 and not multiple_5:
    print("The number {} is a multiple of 7 but not 5".format(num1))
elif not multiple_7 and multiple_5:
    print("The number {} is a multiple of 5 but not 7".format(num1))
elif multiple_5 and multiple_7:
    print("The number {} is a multiple of 7 and 5".format(num1))
else:
    print("The number {} is not a multiple of 7 and 5".format(num1))
````

````python
# Check if a character is uppercase, lowercase, or digit.
char = input("Enter a single character: ")

if char.isupper():
    print("{} is an uppercase letter.".format(char))
elif char.islower():
    print("{} is an lowercase letter.".format(char))
elif char.isdigit():
    print("{} is a digit.".format(char))
else:
    print("{} is a special character or space".format(char))
````

### Edge Cases (This is where most fail)

````python
# ATM withdrawal
num = int(input("Enter withdrawal amount:"))
balance = 10000
withdrawal = (num % 100 == 0)
sufficient_funds = (num <= balance)

if sufficient_funds and withdrawal:
    balance -= num
    print("Rupees {} was debited from you account, Available balance : {}".format(num, balance))
else:
    if not withdrawal:
        print("Transaction declined! Please enter a multiple of 100.")
    else:
        print("Transaction declined! Insufficient funds.")
````
