```python
class Account: 

    def __init__(self, bal, acc): 
        self.bal = bal 
        self.acc = acc

    def debit(self, deb): 
        self.deb = deb
        self.avl_bal = self.bal - self.deb
        print("Your account {} is debited with {} and available balance {}".format(self.acc, self.deb, self.avl_bal))

    def credit(self, cre): 
        self.cre = cre
        self.avl_bal = self.bal + self.cre
        print("Your account {} is debited with {} and available balance {}".format(self.acc, self.cre, self.avl_bal))

a1 = Account(10000, 12345) 
print("Acoount no: {} has balance: {}".format(a1.acc, a1.bal))
a1.credit(5000)
a1.debit(3000)
```
