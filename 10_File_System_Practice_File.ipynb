#### Notes Writer

```python
def save_note(): 
    text = str(input("Enter your note: ")) 
    
    with open('./Notes_Collector/notes.txt', 'w') as file: 
        file.write(text)

    print("Note saved successfully!")

save_note()
```

#### Read Notes 

```python
def read_notes():
    try: 
        with open('./Notes_Collector/notes.txt', 'r') as file: 
            print("Saved Notes: ")
            print(file.read())
    except FileNotFoundError: 
        print("File not found!")
read_notes()
```

#### Daily Journal App

```python
def add_journal_entry(): 
    word = str(input("Enter 'add' if you want to continue adding and 'esc' to stop: "))
    while word == "add": 
        text = input("Enter what you want to add: ")
        with open('./Notes_Collector/notes.txt', 'a') as file:
            file.write("\n----------------\n")
            file.write(text)
        word = str(input("Enter 'add' if you want to continue adding and 'esc' to stop: "))   
```

```python
add_journal_entry()
```
