```python
# Nested conditional statements
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade: A")
    if marks >= 95:
        print("Outstanding!!")
    else:
        print("Excellent!!")
elif marks >= 80: 
    print("Grade : B")
    if marks >= 85: 
        print("Very Good!!")
    else:
        print("Good!!")
else:
    print('Grade: C')
    if marks >= 60:
        print("Average!")
    else:
        print("Below Average!")
```

```python
# Shorthand `if` conditional expression
number = 24

if number % 2 == 0:
    parity = 'even'
else:
    parity = 'odd'
print('The number {} is {}'.format(number, parity))
```

Python provides a shorter syntax, which allows writing such conditions in a single line of code. It is known as a *conditional expression*

x = true_value if condition else false_value

```python
number = 35
parity = 'even' if number % 2 == 0 else 'odd'

print('The number {} is {}'.format(number, parity))
```

### The `pass` statement

`if` statements cannot be empty, there must be at least one statement in every `if` and `elif` block. You can use the `pass` statement to do nothing and avoid getting an error.

```python
number = 27
if number % 2 == 0:
elif number % 3 == 0:
    print("Divisible by 3.")
```

```python
number = 27
if number % 2 == 0:
    pass
elif number % 3 == 0:
    print("Divisible by 3.")
```

## Iteration with `while` loops

Let's try to calculate the factorial of `100` using a `while` loop. The factorial of a number `n` is the product (multiplication) of all the numbers from `1` to `n`, i.e., `1*2*3*...*(n-2)*(n-1)*n`.

```python
result = 1
i = 1

while i <= 100:
    result *= i
    i += 1 

print("The factorial of 100 is {}".format(result))
```

Here's another example that uses two `while` loops to create an interesting pattern.

```python
line = '*'
max_length = 10

while len(line) < max_length:
    print(line)
    line += '*'
while len(line) > 0:
    print(line)
    line = line[:-1]
```

```python
i = 1
j = 10
k = 0
l = 9
while i <= 11:
    print(" " * j + "*" * i + "*" * k + " " * l)
    i += 1
    j -= 1
    k += 1
    l -= 1

i = 10
j = 1
k = 9
l = 0

while i >= 1:
    print(" " * j + "*" * i + "*" * k + " " * l)
    i -= 1
    j += 1
    k -= 1
    l += 1
```

### `break` and `continue` statements

You can use the `break` statement within the loop's body to immediately stop the execution and *break* out of the loop (even if the condition provided to `while` still holds true).

```python
i = 1
result = 1

while i <= 100:
    result *= i
    if i == 42:
        print('Magic number 42 reached! Stopping execution..')
        break
    i += 1
    
print('i:', i)
print('result:', result)
```

```python
i = 1
result = 1

while i < 20:
    i += 1
    if i % 2 == 0:
        print('Skipping {}'.format(i))
        continue
    print('Multiplying with {}'.format(i))
    result = result * i
    
print('i:', i)
print('result:', result)
```
