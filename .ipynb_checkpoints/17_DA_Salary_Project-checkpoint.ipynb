```python
conditions = [
    (df['SalariesReported'] <= 3),
    (df['SalariesReported'] > 3) & (df['SalariesReported'] <= 10), 
    (df['SalariesReported'] > 10)
] 

choices = ['Low', 'Medium', 'High'] 

df['SalaryConfidence'] = np.select(conditions, choices, default = "Unknown")
```
