```python
ypoints = np.array([4, 2, 8, 5, 1, 9]) 
plt.plot(ypoints, marker = 'o')
plt.show()
```

```python
ypoints = np.array([4, 2, 8, 5, 1, 9]) 
plt.plot(ypoints, marker = 'X') # 'o', '+', '*', 'x'
plt.show()
```

#### **Format String** - `fmt` -- `marker | line | color`
* `marker` - `o`, `*`, `.`, `x`, `X`, `+`, `P`, `s`
* `line` - `-`, `:`, `--`, `-.`
* `color` - `r`, `g`, `b`, `c`, `m`, `y`, `k`, `w` 

```python
ypoints = np.array([3, 1, 5, 2, 8, 4])
plt.plot(ypoints, 's:g')
plt.show();
```

```python
# marker size - ms = 20
ypoints = np.array([3, 1, 5, 2, 8, 4])
plt.plot(ypoints, 'X-.g', ms = 15)
plt.show();
```

```python
# marker color - mec , mfc
ypoints = np.array([3, 1, 5, 2, 8, 4])
plt.plot(ypoints, 'X-.g', ms = 15, mec = 'r', mfc = 'c')
plt.show();
```

```python
# linestyle / ls = 'dotted'
ypoints = np.array([3, 1, 5, 2, 8, 4])
plt.plot(ypoints, linestyle = 'dotted', linewidth = 7, color = 'c', marker = 'o', mec = 'g', mfc = 'r', ms = 15)
plt.show();
```

```python
# Multiple Line
y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 11])

plt.plot(y1)
plt.plot(y2)

plt.show()
```

```python
# Labels
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

```python
# Title
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

```python
# Font Properties for Title and Labels
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family':'serif','color':'darkblue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("Sports Watch Data", fontdict = font1, loc = 'left')
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.show()
```

```python
# Grid
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data", loc = 'right')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.grid()
plt.show()
```

```python
# Grid
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data", loc = 'right')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.grid(axis = 'x')
plt.show()
```

```python
# Grid
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data", loc = 'right')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.grid(axis = 'y', ls = '-.')
plt.show()
```

```python
# Grid
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data", loc = 'right')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.grid(ls = '--', linewidth = 0.8, color = 'red')
plt.show()
```

```python
# Subplot

# Plot 1
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10]) 
plt.subplot(1, 2, 1) 
plt.plot(x, y)
plt.title("X vs Y")
plt.xlabel("X - axis")
plt.ylabel("Y - axis")
plt.tight_layout()

# PLot 2 
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40]) 
plt.subplot(1, 2, 2) 
plt.plot(x, y)
plt.title("X vs Y")
plt.xlabel("X - axis")
plt.ylabel("Y - axis")
plt.tight_layout()

plt.show()
```
