## Notes Writer

```python
def save_note(): 
    text = str(input("Enter your note: ")) 
    
    with open('./Notes_Collector/notes.txt', 'w') as file: 
        file.write(text)

    print("Note saved successfully!")

save_note()
```

#### Read Notes 

```python
def read_notes():
    try: 
        with open('./Notes_Collector/notes.txt', 'r') as file: 
            print("Saved Notes: ")
            print(file.read())
    except FileNotFoundError: 
        print("File not found!")
read_notes()
```

## Daily Journal App

```python
def add_journal_entry(): 
    while True:
        word = input("Enter 'add' to continue or 'esc' to stop: ").lower()

        if word == 'esc': 
            print("Program stopped.")
            break 
        elif word == 'add': 
            text = input("Enter what you want to add: ")

            with open('./Notes_Collector/notes.txt', 'a') as file: 
                file.write("\n------------------\n")
                file.write(text)
        else: 
            print("Invalid input.")

add_journal_entry()
```

```python
def read_notes():
    try: 
        with open('./Notes_Collector/notes.txt', 'r') as file: 
            print("Saved Notes: ")
            print(file.read())
    except FileNotFoundError: 
        print("File not found!")
read_notes()
```

## Student Marks Manager

```python
def students_marks_entry(): 
    while True: 
        word = input("Enter 'add' to continue or 'esc' to stop: ").lower()

        if word == 'esc': 
            print("Program stopped.")
            break
        elif word == 'add':  
            roll_no = int(input("Enter student roll no:"))
            name = str(input("Enter student name:"))
            marks = int(input("Enter student marks:"))
            with open('./Students_Marks_Manager/students_marks.txt', 'a') as file: 
                file.write("{}, {}, {}\n".format(roll_no, name, marks))
        else: 
            print("Invalid input.")
```

```python
students_marks_entry()
```

```python
def read_students_data(): 
    try: 
        with open('./Students_Marks_Manager/students_marks.txt', 'r') as file: 
            print("Student Data: ")
            print(file.read())
    except FileNotFoundError: 
        print("File not found!")
```

```python
read_students_data()
```

## CSV Expense Tracker

```python
import csv
```

```python
os.makedirs('./CSV_Expense_Tracker')
```

```python
def add_expense(): 

    while True: 
        choice = input("Enter 'add' to add expense or 'esc' to stop: ").lower()

        if choice == 'esc': 
            print("Program stopped.")
            break

        elif choice == 'add':

            date = input("Enter date (DD-MM-YYYY): ")
            category = input("Enter category: ")
            amount = float(input("Enter amount; "))

            with open('./CSV_Expense_Tracker/expenses.csv', 'a', newline = '') as file: 
                writer = csv.writer(file)
                writer.writerow([date, category, amount])

            print("Expense added successfully!")

        else: 
            print("Invalid input.")
```

```python
add_expense()
```

```python
def view_expenses():
    try: 
        with open('./CSV_Expense_Tracker/expenses.txt', 'r') as file: 
            reader = csv.reader(file)
            print("\n Date | Category | Amount")
            print("-" * 30)

            for row in reader: 
                print(f" {row[0]}, | {row[1]} | {row[2]}")
                
    except FileNotFoundError: 
        print("File not found!")
```

```python
view_expenses()
```

## JSON-Based User Database

```python
import json 
```

```python
os.makedirs('./JSON_Based_User_Database')
```

```python
def add_users(): 
    while True: 
        choice = input("Enter 'add' to add user or 'esc' to stop: ").lower()

        if choice == "esc":
            print("Program stopped.")
            break

        elif choice == 'add': 
            name = input("Enter user name: ")
            age = input("Enter user age: ")

            user_data = {
                "name" : name, 
                "age" : age
            }

            data = {"users" : []}

            data["users"].append(user_data)

            with open("./JSON_Based_User_Database/users.json", "w") as file: 
                json.dump(data, file, indent = 4)

            print("User added successfully!")

        else: 
            print("Invalid input.")
```

```python
add_users()
```

```python
def view_user():
    try: 
        with open("./JSON_Based_User_Database/users.json", "r") as file: 
            data = json.load(file)

            print("\n Users List")
            print("------------------------")

            for user in data["users"]: 
                print(f"Name: {user['name']} | Age: {user['age']}")

    except FileNotFoundError: 
        print("No user file found.")
```

```python
view_user()
```
