# Loops in Python

There are only two you need to master:

1. `for` loop → when you know how many times
2. `while` loop → when you don’t know how many times

### `FOR` LOOP (Structured Iteration)

```python
# Print numbers from 10 to 1.
for i in range(10, 0, -1):
    print(i)
```

```python
# Print all even numbers from 1 to 20
for i in range(0,21, 2):
    print(i)
```

```python
# Find sum of numbers from 1 to 100
counter = 0
for i in range(1,111):
    counter += i
print(counter)
```

```python
# Print multiplication table of a number
counter = 4
for i in range(1,11):
    value = counter * i
    print(value)
```

```python
# Count how many numbers are divisible by 3 (1–50)
counter = 0
for i in range(1,51):
    if i % 3 == 0:
        counter += 1
        
print(counter)
```

```python
# Print numbers from 1 to 20 using a loop
for i in range(1,21):
    print(i)
```

```python
# Print numbers from 20 to 1 (reverse)
for i in range(20, 0, -1):
    print(i)
```

```python
# Print all odd numbers between 1 and 50
for i in range(1,52, 2):
    print(i)
```

### Loop + If (This is where real logic begins)

```python
# Print numbers from 1 to 100 divisible by 3
for i in range(1,101):
    if i % 3 == 0:
        print(i, end = " ")
```

```python
# Count how many numbers between 1–100 are divisible by 5
counter = 0
for i in range(1, 101):
    if i % 5 == 0:
        counter += 1
print("{} numbers between 1-100 are divisible by 5".format(counter))
```

```python
# Print numbers divisible by both 3 and 5
for i in range(1,101):
    if (i % 3 == 0) and (i % 5 == 0):
        print(i, end = " ")
```

```python
# Print squares of numbers from 1 to 10
for i in range(1,11):
    print(i**2, end = " ")
```

```python
# Find the largest number in a list
nums = [12, 45, 7, 89, 23]
largest = nums[0]
for num in nums:
    if num > largest:
        largest = num
print(largest)
```

### Input-Based Loops (Real Thinking Starts)

```python
# Take a number n and print numbers from 1 to n
n = int(input("Enter an integer 'n':"))
for i in range(1, n + 1):
    print(i, end = ' ')
```

```python
# Print multiplication table of a given number
n = int(input("Enter an integer 'n':"))
for i in range(1,11):
    value = n * i
    print(value)
```

```python
# Count digits in a number (e.g., 1234 → 4 digits)
num = int(input("Enter a number: "))
count = 0
num = abs(num)
for _ in str(num):
    count += 1
print("Number of digits:", count)
```

```python
# Reverse a number (123 → 321)
num = int(input("Enter a number: "))
n = abs(num)
rev = 0
for digit in str(n):
    rev = rev * 10 + int(digit)
print("Reveresed number:", rev)
```

### `WHILE` LOOP (Condition-Based)

```python
# Print numbers until user enters 0
i = 1
n = int(input("Enter an integer: "))
while n != 0:
    print(i)
    i += 1
    n = int(input("Enter an integer: "))
```

```python
# Keep taking input until a negative number is entered
n = int(input("Enter an integer: "))
while n > 0:
    n = int(input("Enter an integer: "))
```

```python
# Guess the number game (loop until correct guess)
unkown_number = 9
n = int(input("Enter an integer: "))
while n != unkown_number:
    n = int(input("Enter an integer: "))
print("The number is : ", unkown_number)
```
